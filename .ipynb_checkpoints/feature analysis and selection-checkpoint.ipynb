```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("googleplaystore_clean.csv")
df.info()
```

```python
df.head()
```

```python
num_cols = ["Rating", "Reviews", "Installs", "Price", "Size_MB"]
pd.options.display.float_format = '{:,.2f}'.format
dfDesc = df[num_cols].describe()
dfDesc
```

```python
ratingq1 = dfDesc.loc["25%", "Rating"]
ratingq3 = dfDesc.loc["75%", "Rating"]

ratingIQR = ratingq3-ratingq1

reviewq1 = dfDesc.loc["25%", "Reviews"]
reviewq3 = dfDesc.loc["75%", "Reviews"]

reviewIQR = reviewq3-reviewq1
print("rating IQR =", ratingIQR)
print("reviews IQR =", reviewIQR)
```

```python
df[["Rating", "Reviews", "Installs", "Price", "Size_MB"]].skew()
```

```python
fig, axes = plt.subplots(1,2,figsize= (10,4))
sns.histplot(data = df, x = "Reviews", bins = 50, ax = axes[0])
axes[0].set_title("Reviews")
axes[0].set_xlabel("Number of Reviews")
axes[0].set_ylabel("Number of apps")

sns.histplot(data = df, x = "Rating", bins = 30, ax = axes[1])
axes[1].set_title("Rating")
axes[1].set_xlabel("Rating (Stars)")
axes[1].set_ylabel("Number of Apps")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize = (7,4))
sns.histplot(data = df[df["Reviews"] > 0], x = "Reviews", bins = 50, log_scale = True)
plt.title("Distribution of App Review Counts (log Scale)")
plt.xlabel("Number of Reviews(log scale)")
plt.ylabel("Number of Apps")
plt.show()
```

```python
df["Log_Reviews"] = np.log10(df["Reviews"] + 1)
df["Log_Installs"] = np.log10(df["Installs"] + 1)
df["Log_Price"] = np.log10(df["Price"] + 1)

df[["Reviews", "Log_Reviews", "Installs", "Log_Installs", "Price", "Log_Price"]].skew()
```

```python
df.head()
```

```python
df["Is_Paid"] = (df["Type"] == "Paid").astype(int)
df["Is_Paid"].value_counts(normalize = True) * 100
```

```python
(df["Is_Paid"] == (df["Price"] > 0)).mean()
```

```python
df["Last Updated"] = pd.to_datetime(df["Last Updated"])
ref_date = df["Last Updated"].max() #Single reference date that is the latest (one date of an app)
df["Days_Since_Update"] = (ref_date - df["Last Updated"])

print(ref_date)
print(df["Days_Since_Update"].dtype)
df["Days_Since_Update"].describe()
```

```python
df.head()
```

```python
df["Days_Since_Update"] = (ref_date - df["Last Updated"]).dt.days
df["Days_Since_Update"].dtype
```

```python
cols = ["Days_Since_Update", "Log_Reviews", "Log_Price", "Is_Paid", "Size_MB", "Log_Installs"]
rated = df[df["Rating"].notna()]      

rated[cols].corrwith(rated["Rating"], method="spearman").sort_values()
```

```python
df["Days_Since_Update"].dtype
```

```python
df["Size_Varies_int"] = df["Size_Varies"].astype(int)

cols2 = ["Log_Reviews", "Days_Since_Update", "Log_Price", "Is_Paid", "Size_MB", "Size_Varies_int"]
df[cols2].corrwith(df["Log_Installs"], method="spearman").sort_values()
```

```python
df.groupby("Size_Varies")["Installs"].median()
```

```python
df.groupby("Type")["Installs"].median()
```

```python
heat_cols = ["Rating", "Log_Reviews", "Log_Installs", "Log_Price", "Is_Paid", "Size_MB", "Size_Varies_int", "Days_Since_Update"]

rated = df[df["Rating"].notna()]
print("rated apps:", len(rated)) 

corr = rated[heat_cols].corr(method="spearman")
print(corr.round(2))
plt.figure(figsize=(9, 7))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm",
            center=0, vmin=-1, vmax=1, square=True)
plt.title("Spearman correlation (rated apps, n = 8,196)")
plt.tight_layout()
plt.show()
```

```python
#RQ2

rs = df[df["Rating"].notna() & df["Size_MB"].notna()].copy()
print("apps in chart:", len(rs))

bins = [0,5,10,20,50,100]
rs["Size_Band"] = pd.cut(rs["Size_MB"], bins = bins, include_lowest = True)
print(rs["Size_Band"].value_counts().sort_index())

plt.figure(figsize = (8,5))
sns.boxplot(data = rs, x = "Size_Band", y = "Rating")
plt.title("Rating by app size band")
plt.xlabel("Size (MB)")
plt.ylabel("Rating (stars)")
plt.tight_layout()
plt.show()
```

```python
rs.groupby("Size_Band", observed = True)["Rating"].median()
```

```python
cat = rated.groupby("Category")["Rating"].agg(["count", "median", "mean"])
cat = cat[cat["count"] >=100].sort_values("median")
print(cat)

plt.figure(figsize=(8, 8))
plt.barh(cat.index, cat["median"])
plt.xlim(3.5, 5)
plt.title("Median rating by category (categories with 100+ rated apps)")
plt.xlabel("Median rating")
plt.tight_layout()
plt.show()
```

```python
#Testing version features
df["Min_Android"] = df["Android Ver"].str.extract(r"(\d+\.?\d*)")[0].astype(float)

print("Current Ver unique values:", df["Current Ver"].nunique(), "out of", len(df))
df["Major_Ver"] = df["Current Ver"].str.extract(r"(\d+)")[0].astype(float)

rated = df[df["Rating"].notna()]

print("\nvs Rating")
print(rated[["Min_Android", "Major_Ver"]].corrwith(rated["Rating"], method = "spearman").round(2))

print("\nvs Log_Installs")
print(df[["Min_Android", "Major_Ver"]].corrwith(df["Log_Installs"], method = "spearman").round(2))
```
