```python
df = df_raw.copy()
log = [("raw data", len(df))]

df = df[df["Category"] != "1.9"] #corrupt row with the strange category.
log.append(("corrupt row removed", len(df)))

df.shape
```

```python
df.isna().sum()
```

```python
df.dtypes
```

```python
df["Reviews"] = df["Reviews"].astype(int)
df["Installs"] = df["Installs"].str.replace('[+,]',"", regex = True).astype(int) #.str to be able to use replace from it to manipulate the individual strings. and remove all + and , using the replace method.
df["Price"] = df["Price"].str.replace("$", "", regex = False).astype(np.float64) #regex = false because it would interpret $ not as $ but as end of the text

def convertSize(s):
    if s.endswith("M"):
        return float(s[:-1])
    if s.endswith("k"):
        return float(s[:-1]) / 1024 #one megabyte is 1024 kilobyte
    return np.nan #for varies with device value

df["Size_MB"] = df["Size"].apply(convertSize)
df["Last Updated"] = pd.to_datetime(df["Last Updated"])

df.info()
```

```python
df["Size_MB"].isna().sum()
```

```python
df["Price"].max()
```

```python
df = df.drop_duplicates()
log.append(("exact duplicates removed", len(df)))
len(df)
```

```python
df = (df.sort_values(["Last Updated", "Reviews"], ascending = False).drop_duplicates(subset = "App", keep = "first"))
log.append(("duplicate apps handled based on last updated", len(df)))
pd.DataFrame(log, columns=["step", "Number of rows"])
```

```python
df.isnull().sum()
```

```python
df[df["Type"].isna()][["App", "Type", "Price","Installs"]]
```

```python
df[(df["Type"] == "Free") & (df["Price"] != 0)] #No apps that has some price but is marked free
```

```python
df[(df["Type"] == "Paid") & (df["Price"] == 0)] #No apps that is paid and has a price 0
```

```python
df.loc[df["Type"].isna(), "Type"] = "Free" #setting the value of where the row is nan in Type column
df["Type"].value_counts()
```

```python
s = df["Size_MB"].dropna()

q1 = s.quantile(0.25)
q3 = s.quantile(0.75)

iqr = q3 - q1
upper = q3 + 1.5*iqr
lower = q1 - 1.5*iqr

print("outliers above: ", (s > upper).sum(), "out of", len(s))
print("outliers below: ", (s < lower).sum(), "out of", len(s))
print("mean:",s.mean(), "\nmedian:", s.median(), "\nmax:", s.max())
```

```python
df["Size_Varies"] = df["Size"] == "Varies with device"
size_median = df["Size_MB"].median()
print("median used for fill:", size_median)

df["Size_MB_Filled"] = df["Size_MB"].fillna(size_median)

print(df["Size_Varies"].sum())
print(df["Size_MB"].isna().sum())
print(df["Size_MB_Filled"].isna().sum())
df.shape
```

```python
df.columns
```

```python
print("correlation between rating and size without imputation:", df["Size_MB"].corr(df["Rating"]))
print("correlation between rating and size with imputation:", df["Size_MB_Filled"].corr(df["Rating"]))
```

```python
rated = df.dropna(subset = ["Rating"])
print(len(rated)) #apps with rating
```

```python
#logically odd values
paid = df.loc[df["Price"] > 0, "Price"]
q1_paid = paid.quantile(0.25)
q3_paid = paid.quantile(0.75)

extreme = q3_paid + 3 * iqr
print("Extreme limit:", extreme)
print(df[df["Price"] > extreme][["App", "Price"]].sort_values("Price", ascending = False))
print()
print((df["Reviews"] > df["Installs"]).sum())
```

```python
log.append(("Final cleaned dataset", len(df)))
display(pd.DataFrame(log,columns = ["Step", "number of rows"]))
df.to_csv("googleplaystore_clean.csv", index = False)
print(df["Rating"].between(1,5).sum(), "of", df["Rating"].notna().sum())
```

# Visualizations


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('googleplaystore_clean.csv', parse_dates=['Last Updated'])
df.head()
```

```python
# Log versions for skewed columns (used in some charts)
df['LogReviews'] = np.log10(df['Reviews'] + 1)
```

```python
df['LogInstalls'] = np.log10(df['Installs'].clip(lower=1))
```

```python
# Rating Histogram
plt.figure(figsize = (8,5))
sns.histplot(df['Rating'], bins=30, kde=True)
plt.title('Rating Distribution')
plt.xlabel('Rating')
plt.ylabel('Number of Apps')
plt.show()
```

```python
# No. of apps by category
top = df['Category'].value_counts().head(15)
plt.figure(figsize = (9, 6))
sns.barplot(x = top.values, y = top.index, color = 'steelblue')
plt.title('Number of Apps by Category Top(15)')
plt.xlabel('Number of apps')
plt.ylabel('Category')
plt.show()
```

```python
# Rating by Category (Boxplot)
top10 = df['Category'].value_counts().head(10).index
plt.figure(figsize = (10, 6))
sns.boxplot(data = df[df['Category'].isin(top10)], 
            x = 'Rating', y = 'Category', 
            order=top10)
plt.title('Rating by category (10 largest)')
plt.show()
```

```python
# Average rating by category
avg = df.groupby('Category')['Rating'].mean().sort_values()
plt.figure(figsize = (9, 9))
sns.barplot(x = avg.values, y = avg.index, color = 'seagreen')
plt.xlim(3.5, 4.6)  # Starts at 3.5 so small differences are visible
plt.title('Average Rating by Category')
plt.xlabel('Average rating')
plt.ylabel('Category')
plt.show()
```

```python
# Free vs paid app
plt.figure(figsize = (6, 5))
ax = sns.countplot(data = df, x = 'Type', 
                   color = 'lightgreen',
                   order = ['Free', 'Paid'])
ax.bar_label(ax.containers[0])  # shows the number on each bar
plt.title('Free vs Paid App Count')
plt.xlabel('Type')
plt.ylabel('Number of apps')
plt.show()
```

```python
# Free vs Paid Rating (Boxplot)
plt.figure(figsize = (6, 5))
sns.boxplot(data = df, x = 'Type', y = 'Rating', 
            order = ['Free', 'Paid'], 
            palette=sns.color_palette(n_colors=2))
plt.title('Rating: Free vs Paid')
plt.show()

print(df.groupby('Type')['Rating'].agg(['mean', 'median']).round(2))
```

```python
# Free vs Paid Apps by Category
top10 = df['Category'].value_counts().head(10).index
subset = df[df['Category'].isin(top10)]

plt.figure(figsize = (10, 6))
sns.countplot(data = subset, y = 'Category', hue = 'Type', order = top10)
plt.title('Free vs Paid Apps by Category (top 10)')
plt.xlabel('No. of Apps')
plt.show()
```

```python
# Categories with the highest share of paid apps
share = (pd.crosstab(df['Category'], df['Type'], 
                     normalize='index') * 100)['Paid'].sort_values(ascending=False).head(10)

plt.figure(figsize=(9, 5))
sns.barplot(x=share.values, y=share.index, color='coral')
plt.title('Categories with the Highest Share of Paid Apps')
plt.xlabel('% of apps that are paid')
plt.show()
```

```python
# Size vs Rating (Scatter plot)
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Size_MB', y='Rating', alpha=0.25, s=12)
plt.title('App Size vs Rating')
plt.xlabel('Size (MB)')
plt.ylabel('Rating')
plt.show()
```

```python
# Review volume: Free vs Paid (Boxplot, log scale)
plt.figure(figsize=(6, 5))
sns.boxplot(data=df, x='Type', y='LogReviews', order=['Free', 'Paid'])
plt.title('Review Volume: Free vs Paid')
plt.ylabel('Reviews (log10)')
plt.show()

print(df.groupby('Type')['Reviews'].median())
```

```python
# Installs vs Rating (Scatter Plot)
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Installs', y='Rating', alpha=0.3)
plt.xscale('log')
plt.title('Installs vs Rating')
plt.xlabel('Installs (log scale)')
plt.ylabel('Rating')
plt.show()
```

```python
# Installs vs Reviews (Scatter plot)
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Installs', y='Reviews', alpha=0.3)
plt.xscale('log'); plt.yscale('log')
plt.title('Installs vs Reviews')
plt.xlabel('Installs (log scale)')
plt.ylabel('Reviews (log scale)')
plt.show()
```

```python
# Total installs by category (Top 10)
inst = (df.groupby('Category')['Installs'].sum()
          .sort_values(ascending=False)
          .head(10) / 1e9)            # convert to billions

plt.figure(figsize=(9, 5))
sns.barplot(x=inst.values, y=inst.index, color='mediumpurple')
plt.title('Total Installs by Category (Top 10)')
plt.xlabel('Installs (billions)')
plt.ylabel('Category')
plt.show()

print(inst.round(1))
```
